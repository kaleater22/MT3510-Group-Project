```python
import itertools as it
import networkx as nx
import matplotlib.pyplot as plt

D = nx.MultiDiGraph() # Use MultiDiGraph to draw

# a transitions
D.add_edge(0,1,label='a')
D.add_edge(1,2,label='a')
D.add_edge(2,2,label='a')

# b transitions
D.add_edge(0,0,label='b')
D.add_edge(1,0,label='b')
D.add_edge(2,1,label='b')

fig, ax = plt.subplots()

connectionstyle = [f"arc3,rad={r}" for r in it.accumulate([0.15]*4)] # Works with arc3 and angle3 connectionstyles

pos = nx.shell_layout(D)

# Plot nodes with different properties 
# Start state is a red circle
nx.draw_networkx_nodes(D, pos, node_size=300, nodelist=[1, 2], node_color="tab:blue")
nx.draw_networkx_nodes(D, pos, node_size=300, nodelist=[0], node_color="tab:red")
nx.draw_networkx_nodes(D, pos, node_size=200, node_color="yellow", ax=ax)

nx.draw_networkx_labels(D, pos, font_size=15, ax=ax)
nx.draw_networkx_edges(D, pos, edge_color="grey", connectionstyle=connectionstyle, arrows=True, ax=ax)


# labelling the edges
labels = {
    (u,v,k): d["label"]
    for u,v,k,d in D.edges(keys=True, data=True)
}

nx.draw_networkx_edge_labels(
    D,
    pos,
    edge_labels=labels,
    connectionstyle=connectionstyle,
    label_pos=0.3,
    font_color="blue",
    bbox={"alpha":0},
    ax=ax
)

plt.axis("off")
plt.show()
```
